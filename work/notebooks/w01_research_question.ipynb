## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Question:** Which pages should a content team review first this week, out of everything in their inventory?

**Decision it improves:** A content/SEO team has limited review capacity (say, 20-50 pages/week) and needs to know which pages are worth a human's time versus which are fine as-is.

**Who acts on it:** A content strategist or SEO analyst, using my ranked queue as their weekly to-do list.

**The action:** For each flagged page, the reviewer checks the reason code (e.g. "stale_visible_page" or "declining_with_demand") and decides on a specific fix — refresh the content, rewrite for a keyword gap, or leave it and monitor.

**Cost of a wrong call:**
- *False positive* (flagged as worth reviewing, but it wasn't): wastes a reviewer's limited time on a page that didn't need attention — the real cost is opportunity cost, since a genuinely declining page waited another week.
- *False negative* (a real problem page never surfaces): the page keeps losing visibility unnoticed, and by the time it's caught the drop may be harder to reverse.
Because review capacity is the scarce resource, precision in the top of the ranked list (Precision@K) matters more than catching every possible case.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Loading the starter dataset to check whether "declining" pages are common enough, and visible enough, to be worth 7 weeks of work.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/MaryumAkram16/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Number 1: how many pages are currently flagged as declining
declining = df[df["trend_direction"] == "down"]
print(f"Declining pages: {len(declining)} of {len(df)} total ({len(declining)/len(df):.1%})")

# Number 2: of those, how many are also stale (a likely-actionable overlap)
stale_declining = declining[declining["days_since_last_update"] >= 180]
print(f"Declining AND stale (>=180 days since last update): {len(stale_declining)} "
      f"({len(stale_declining)/len(declining):.1%} of declining pages)")

# Number 3: how much visible traffic is sitting on declining pages
total_impressions_declining = declining["impressions_90d"].sum()
total_impressions_all = df["impressions_90d"].sum()
print(f"Share of all 90-day impressions sitting on declining pages: "
      f"{total_impressions_declining/total_impressions_all:.1%}")
```

**What this shows:** A striking share of the inventory — over half — is currently declining, and those pages carry roughly half of all 90-day impressions, so this is not a "long tail nobody sees" problem; real visibility is at stake. What's more interesting is what the overlap check reveals: only 0.5% of declining pages are also stale (updated 180+ days ago). This means the simple story — "old pages decline, refresh them" — does not hold up in this data. Most declining pages were updated relatively recently, which suggests the cause of decline here is something other than neglect (position loss, CTR drop, or competitive pressure are more likely candidates worth investigating). Combined with the ML-01 result (baseline Precision@50 = 0.240 vs. random forest = 0.740, per `docs/ml-intern-dataset-and-lane-guide.md`), there's clearly enough signal in the data to model — but the "stale page" reason code alone won't explain most of it, which is itself a useful early finding to carry into Week 2.
